```python
from faker import Faker
import random

fake = Faker()

def generate_sa_id():
    birth_date = fake.date_of_birth(minimum_age=18, maximum_age=70)
    id_date_part = birth_date.strftime("%y%m%d")
    random_digits = "".join([str(random.randint(0, 9)) for _ in range(7)])
    return id_date_part + random_digits

sample_text = f"""
My name is {fake.first_name()} {fake.last_name()} and my ID number is {generate_sa_id()}.
You can reach me at {fake.email()} or 0{random.randint(60, 89)} {random.randint(100, 999)} {random.randint(1000, 9999)}.
My address is {random.randint(1, 100)} {fake.street_name()}, {fake.city()}, {random.randint(1000, 9999)}.
My credit card number is {fake.credit_card_number(card_type=None)}.
"""

print(sample_text)
```
